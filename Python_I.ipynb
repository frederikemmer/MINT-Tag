## 4) Bedingungen (`if/elif/else`)
Mit Bedingungen reagiert dein Programm unterschiedlich je nach Eingabe.


```python

zahl = int(input("Gib eine ganze Zahl ein: "))
if zahl % 2 == 0:
    print("Gerade Zahl ✅")
else:
    print("Ungerade Zahl ❗")
```

### Aufgabe 4
Schreibe ein Programm, das eine Schulnote (1–6) einliest und eine **Wortbeschreibung** ausgibt:
- 1 = "sehr gut", 2 = "gut", 3 = "befriedigend", 4 = "ausreichend", 5 = "mangelhaft", 6 = "ungenügend".


```python

# Deine Lösung hier
note = int(input("Note (1-6): "))
if note == 1:
    print("sehr gut")
elif note == 2:
    print("gut")
elif note == 3:
    print("befriedigend")
elif note == 4:
    print("ausreichend")
elif note == 5:
    print("mangelhaft")
elif note == 6:
    print("ungenügend")
else:
    print("Ungültige Note.")
```

## 5) Schleifen & Listen
Eine Liste kann mehrere Werte enthalten. Mit `for` kannst du wiederholt etwas tun.


```python

lieblingsgerichte = ["Pizza", "Sushi", "Käsespätzle"]
for gericht in lieblingsgerichte:
    print("Ich mag:", gericht)
```

### Aufgabe 5a
Erstelle eine Liste mit **3 Lieblingsfächern** und gib alle nacheinander aus.

### Aufgabe 5b (Challenge)
Gib die Zahlen von 1 bis 20 aus. Bei **Vielfachen von 3** gib „Fizz“ aus, bei **Vielfachen von 5** „Buzz“, bei **Vielfachen von 3 und 5** „FizzBuzz“.


```python

# Deine Lösung hier
for n in range(1, 21):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

## 6) Mini‑Challenge (zum Abschluss)
Schreibe ein **Mini-Programm** mit `input()` und `if`, das eine Entscheidung trifft.
Beispiele: „Darf ich den Film schauen?“ (ab 12, 16, 18), „Ist die Zahl größer als 100?“ etc.


```python

# Deine eigene Mini-Challenge
schwelle = 100
zahl = int(input("Gib eine ganze Zahl ein: "))
if zahl > schwelle:
    print("Die Zahl ist größer als", schwelle)
else:
    print("Die Zahl ist nicht größer als", schwelle)
```

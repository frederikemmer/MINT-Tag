## 1) Variablen

In jeder Programmiersprache arbeitet man mit Variablen.
Diese können verschiedene Datentypen haben und in Funktionen verwendet werden.

```python
# String (Text) "str"
a = "Hallo Welt"

# Integer (Ganzzahl) "int"
b = 42

# Float (Kommazahl) "float"
c = 3.14

# Boolean (Wahrheitswert) "bool"
d = True

# Function (Funktion)
print(a)
print(b)
print(c)
print(d)
```

```python
# Ints / Floats können für mathematische Operationen verwendet werden
a = 5
b = 3
y = a + b

print(y)
```


---------
Der Datentyp wird automatisch ausgewählt und Python hilft hier an manchen Stellen etwas mit.
Andere Programmiersprachen würden hier "false" ausgebeben

```python
a = 1 + 2
b = 3.0

print(a)
print(b)
print(a == b)
```


---
Variablen können aber auch für andere Zwecke verwendet werden

```python
# Variablen können aber auch für andere Zwecke verwendet werden
a = "MINT-Tag"
b = "Campus Burghausen"
y = a + " am " + b

print(y)
```

## 4) Bedingungen (`if/elif/else`)
Mit Bedingungen reagiert dein Programm unterschiedlich je nach Eingabe.


```python

zahl = int(input("Gib eine ganze Zahl ein: "))
if zahl % 2 == 0:
    print("Gerade Zahl ✅")
else:
    print("Ungerade Zahl ❗")
```

## 5) Schleifen & Listen

### >>> Aufgabe 5b (Challenge)
Gib die Zahlen von 1 bis 20 aus. Bei **Vielfachen von 3** gib „Fizz“ aus, bei **Vielfachen von 5** „Buzz“, bei **Vielfachen von 3 und 5** „FizzBuzz“.


```python

# Deine Lösung hier
for n in range(1, 21):
    if n % 15 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

## 6) eigene Ideen (zum Abschluss)
Schreibe ein **Mini-Programm** mit zB. `input()` und `if`, das eine Entscheidung trifft oder dir etwas ausrechnet.
Beispiele: „Darf ich den Film schauen?“ (ab 12, 16, 18), „Ist die Zahl größer als 100?“ etc.


```python
# Eingabe: n‑te Fibonacci‑Zahl (n >= 0)
n = int(input("Gib die Position ein (n): "))

# Fibonacci‑Berechnung
a, b = 1, 1
for _ in range(n):
    a, b = b, a + b
    print(a)

print(f"Die {n}‑te Fibonacci‑Zahl ist: {a}")
```

```python

# Deine eigene Mini-Challenge
schwelle = 100
zahl = int(input("Gib eine ganze Zahl ein: "))
if zahl > schwelle:
    print("Die Zahl ist größer als", schwelle)
else:
    print("Die Zahl ist nicht größer als", schwelle)
```

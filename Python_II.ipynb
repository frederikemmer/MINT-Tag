## 2) Mini‑Projekt A: Zahlenraten (mit `random`)
Der Computer wählt eine zufällige Zahl. Du hast mehrere Versuche, sie zu erraten.


```python

import random

def zahlenraten(min_zahl=1, max_zahl=50, versuche=6):
    geheime_zahl = random.randint(min_zahl, max_zahl)
    print(f"Ich habe mir eine Zahl zwischen {min_zahl} und {max_zahl} gedacht.")
    print(f"Du hast {versuche} Versuche. Viel Erfolg!")
    for runde in range(1, versuche + 1):
        t = input(f"Versuch {runde}: Deine Zahl? ")
        if not t.strip().isdigit():
            print("Bitte eine ganze Zahl eingeben.")
            continue
        t = int(t)
        if t < geheime_zahl:
            print("Zu klein!")
        elif t > geheime_zahl:
            print("Zu groß!")
        else:
            print("🎉 Treffer! Du hast gewonnen.")
            return
    print("Schade, verloren! Die Zahl war:", geheime_zahl)

zahlenraten()
```

### Challenge
- Schwierigkeitsgrade (leicht/mittel/schwer) mit unterschiedlichen Zahlenbereichen/Versuchen.
- Zähle die **benötigten Versuche** und zeige am Ende eine Bewertung.


```python

# Beispiel-Lösungsidee (anpassbar)
def zahlenraten_mit_level():
    print("Wähle einen Schwierigkeitsgrad: leicht / mittel / schwer")
    level = input("> ").strip().lower()
    if level == "leicht":
        p = (1, 20, 8)
    elif level == "mittel":
        p = (1, 50, 6)
    else:
        p = (1, 100, 5)
    min_z, max_z, tries = p
    import random
    ziel = random.randint(min_z, max_z)
    anzahl = 0
    print(f"Rate die Zahl zwischen {min_z} und {max_z}. Du hast {tries} Versuche.")
    while anzahl < tries:
        anzahl += 1
        t = input(f"Versuch {anzahl}: ")
        if not t.strip().isdigit():
            print("Bitte eine ganze Zahl!")
            continue
        t = int(t)
        if t == ziel:
            print("🎉 Treffer in", anzahl, "Versuchen!")
            return
        print("Zu klein!" if t < ziel else "Zu groß!")
    print("Leider nicht geschafft. Die Zahl war:", ziel)

zahlenraten_mit_level()
```

## 3) Mini‑Projekt B: Quiz
Ein kleines Multiple‑Choice‑Quiz mit Auswertung.


```python

fragen = [
    {
        "text": "Welches Symbol steht in Python für die Potenz?",
        "optionen": ["^", "**", "pow"],
        "loesung": 2  # 1-basiert: 2 => "**"
    },
    {
        "text": "Wie heißt die Funktion für Eingaben in Python?",
        "optionen": ["input()", "scan()", "read()"],
        "loesung": 1
    },
    {
        "text": "Was bedeutet der Operator % bei ganzen Zahlen?",
        "optionen": ["Division", "Rest der Division", "Multiplikation"],
        "loesung": 2
    }
]

def quiz_spielen(fragen_liste):
    print("Willkommen zum Python-Quiz!")
    punkte = 0
    for i, f in enumerate(fragen_liste, start=1):
        print(f"\nFrage {i}: {f['text']}")
        for idx, opt in enumerate(f["optionen"], start=1):
            print(f"  {idx}) {opt}")
        antwort_text = input("Deine Antwort (1/2/3): ")
        if antwort_text.strip() not in {"1","2","3"}:
            print("Ungültige Eingabe, zählt als falsch.")
            continue
        antwort = int(antwort_text)
        if antwort == f["loesung"]:
            print("Richtig! ✅")
            punkte += 1
        else:
            print("Falsch. ❌")
    print(f"\nFertig! Du hast {punkte} von {len(fragen_liste)} Punkten.")

quiz_spielen(fragen)
```

### Aufgabe (Erweiterung)
- Füge **weitere Fragen** hinzu.
- Vergib am Ende eine **Note** abhängig von der Punktzahl (z. B. ab 80 % = „sehr gut“).


```python

# Beispiel für eine Note nach Prozenten
def note_aus_prozent(punkte, gesamt):
    if gesamt == 0:
        return "—"
    prozent = punkte / gesamt
    if prozent >= 0.9:
        return "1 (sehr gut)"
    elif prozent >= 0.75:
        return "2 (gut)"
    elif prozent >= 0.6:
        return "3 (befriedigend)"
    elif prozent >= 0.45:
        return "4 (ausreichend)"
    elif prozent >= 0.3:
        return "5 (mangelhaft)"
    else:
        return "6 (ungenügend)"

# Beispiel-Aufruf:
print(note_aus_prozent(7, 10))
```

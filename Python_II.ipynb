## 2) Mini‑Projekt A: Zahlenraten (mit `random`)
Der Computer wählt eine zufällige Zahl. Du hast mehrere Versuche, sie zu erraten.


```python
import random

def zufallszahl(von, bis):
    return random.randint(von, bis)

von = int(input("Zufallszahl von: "))
bis = int(input("Zufallszahl bis: "))
print("Zufallszahl: ", zufallszahl(von, bis))
```

```python
import random

def zahlenraten(min_zahl=1, max_zahl=50, versuche=6):
    geheime_zahl = random.randint(min_zahl, max_zahl)
    print(f"Ich habe mir eine Zahl zwischen {min_zahl} und {max_zahl} gedacht.")
    print(f"Du hast {versuche} Versuche. Viel Erfolg!")
    for runde in range(1, versuche + 1):
        t = int(input(f"Versuch {runde}: Deine Zahl? "))
        if t < geheime_zahl:
            print("Zu klein!")
        elif t > geheime_zahl:
            print("Zu groß!")
        else:
            print("🎉 Treffer! Du hast gewonnen.")
            return
    print("Schade, verloren! Die Zahl war:", geheime_zahl)

zahlenraten()
```

### Challenge
- Schwierigkeitsgrade (leicht/mittel/schwer) mit unterschiedlichen Zahlenbereichen/Versuchen.
- Zähle die **benötigten Versuche** und zeige am Ende eine Bewertung.


```python
def zahlenraten_mit_level():
    print("Wähle einen Schwierigkeitsgrad: leicht / mittel / schwer")
    level = input("> ").strip().lower()
    if level == "leicht":
        p = (1, 20, 8)
    elif level == "mittel":
        p = (1, 50, 6)
    else:
        p = (1, 100, 5)
    min_z, max_z, tries = p
    import random
    ziel = random.randint(min_z, max_z)
    anzahl = 0
    print(f"Rate die Zahl zwischen {min_z} und {max_z}. Du hast {tries} Versuche.")
    while anzahl < tries:
        anzahl += 1
        t = input(f"Versuch {anzahl}: ")
        if not t.strip().isdigit():
            print("Bitte eine ganze Zahl!")
            continue
        t = int(t)
        if t == ziel:
            print("🎉 Treffer in", anzahl, "Versuchen!")
            return
        print("Zu klein!" if t < ziel else "Zu groß!")
    print("Leider nicht geschafft. Die Zahl war:", ziel)

zahlenraten_mit_level()
```

## 3) Beispiel für Notenvergabe:


```python
def note_von_versuchen(genutzte_versuche, moegliche_versuche):
    
    prozent = genutzte_versuche / moegliche_versuche
    
    if prozent <= 0.3:
        note = "1 (sehr gut)"
    elif prozent <= 0.45:
        note = "2 (gut)"
    elif prozent <= 0.6:
        note = "3 (befriedigend)"
    elif prozent <= 0.75:
        note = "4 (ausreichend)"
    elif prozent <= 0.9:
        note = "5 (mangelhaft)"
    else:
        note = "6 (ungenügend)"

    return note

note_von_versuchen(4, 10)
```
